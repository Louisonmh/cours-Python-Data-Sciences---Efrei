# Data Preprocessing

```python
#check missing values
titanic.isna().sum()
```

```python
# manage missing values as you see pertinent
titanic = titanic.dropna(subset=['deck'])
```

```python
#check missing values
titanic.isna().sum()
```

```python
# manage missing values as you see pertinent
titanic = titanic.fillna({'age': titanic['age'].median(), 'embarked': titanic['embarked'].mode()})
```

```python
# manage missing values as you see pertinent
titanic.isna().sum()
```

```python
# manage missing values as you see pertinent
titanic = titanic.dropna(subset=['embark_town'])
```

```python
#check missing values
titanic.isna().sum()
```

```python
# Check for duplicate rows
titanic.duplicated().sum()
```

```python
#Drop duplicates keeping the first occurance
titanic = titanic.drop_duplicates(keep='first')
```

```python
# Check for duplicate rows
titanic.duplicated().sum()
```

```python
#show the final dataset 5 first rows
titanic.head()
```

### Indexing

### Using .loc[] (Label-based indexing)

```python
#Select all information about the passenger at index 0.
titanic.loc[titanic.index[0]]
```

```python
#Select all information about the passenger at index 50.
titanic.loc[titanic.index[50]]
```

```python
#Get the sex and age of the passenger at index 10.
titanic.loc[10, ['sex', 'age']]
```

```python
#Select passengers from index 100 to 105 and show only their survived and pclass columns.
titanic.loc[100:105, ['survived', 'pclass']]

```

```python
#Get all passengers who embarked from "Southampton" and display only sex, age, and fare
titanic.loc[titanic['embark_town'] == 'Southampton', ['sex', 'age', 'fare']]
```

```python
#Select the first 5 female passengers and display their age, pclass, and survived.
titanic.loc[titanic['sex'] == 'female', ['age', 'pclass', 'survived']].head(5)
```

### Using .iloc[] (Position-based indexing)

```python
#Display the first row of the dataset.
titanic.iloc[0]
```

```python
#Get the values of age and fare for the 5th passenger (row index 4).
titanic.iloc[4, [3, 6]]
```

```python
#Display the first 10 passengers (rows 0–9) but only the first 3 columns (survived, pclass, sex).
titanic.iloc[0:10, 0:3]
```
