Exercice 2 :

```python
code = 2037
user = int(input("Entre un code :"))

while user != code:
    print("Code incorrect")
    user = int(input("Entre un code :"))
print("Code correct")
```

Exercice 3 :

```python
user = int(input("Enter votre age :"))

for user in range(user,20):
    print("Vous avez", user, "ans")
print("La boucle est terminée")
```

Exercice 4 :

```python
cpt = int(input("Entre un nombre :"))
for cpt in range(cpt, 0, -1):
    print(cpt)
print("La boucle est terminée")
```

Exercice 5 : A

```python
code = 2037
user = int(input("Entre un code :"))

while True:
    if user == code:
        print("Code correct")
        break
    else:
        print("Code incorrect")
        user = int(input("Entre un code :"))
```

Exercie 5 : B

```python
code = 2037

while True:
    user = int(input("Entre un code :"))
    if user != code:
        print("Code incorrect")
        continue        
    print("Code correct")
    break
```
