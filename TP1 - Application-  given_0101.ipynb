**Create a function that grabs the email website domain from a string in the form:**

    user@domain.com
    
**example: "user@domain.com" returns: domain.com**

```python
# Votre code
def domainmail(mail):
    result = []
    for i in range(len(mail)):
        if mail[i]=='@':
            result=mail[i+1:len(mail)]
    return result
```

```python
# Call the function

#Votre code
mail = "user@domain.com"

domainmail(mail)
```

**Create a function that counts the number of times the word "données" occurs in a string. Again ignore edge cases.**

```python
st="Un Data Scientist : il analyse et interprète des données complexes.  Ce sont des chercheurs de données qui organisent des Données Massives."
```

```python
def countData(st):
    # Votre code
    st = st.lower()
    liste = st.split()
    result = 0
    for i in range(len(liste)):
        if liste[i]=='données':
            result+=1
    return result
    
    
```

```python
countData(st)
```

**Use lambda expressions and the filter() function to filter out words from a list that don't start with the letter 's'.** 

**Example:**

    seq = ['soup','dog','salad','cat','great']

**should be filtered down to:**

    ['soup','salad']

```python
# Votre code
seq = ['soup','dog','salad','cat','great']
```

```python
# Votre code
list(filter(lambda var: var[0]=='s', seq))
```

**FizzBuzz:**

**Écrire un programme qui affiche les nombres entiers de 1 à 100. Pour les multiples de trois afficher "Fizz" à la place du nombre et pour les multiples de cinq afficher "Buzz". Pour les nombres qui sont à la fois des multiples de trois et de cinq afficher "FizzBuzz".**

```python
# Votre code
for i in range(1,101):
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    else:
        print(i)

```

**Utilisez la compréhension de liste pour créer une liste avec les 3 premières lettres de chaque mot dans la chaine ci dessous :**

```python
st = "Création d'une liste avec les trois premières lettres de chaque mot de cette chaine de caractères"
```

```python
# Votre code
def lettre3 (mot):
    return mot[0:3]

liste = st.split()
res =[]
res.append(list(map(lettre3, liste)))
print(res)
```
